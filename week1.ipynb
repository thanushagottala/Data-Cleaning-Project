# Data Cleaning & Visualization Project
This notebook contains the data preprocessing, cleaning, and visual report for week 1.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for visualizations
sns.set_theme(style="whitegrid")

# 1. LOAD THE DATA
df = pd.read_csv('sample_data.csv')
print("--- Initial Raw Data ---")
print(df)
print("\n--- Initial Data Info ---")
print(df.info())

# 2. DATA CLEANING
# Step A: Remove complete duplicates
df.drop_duplicates(inplace=True)

# Step B: Handle missing values in numeric columns
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Quantity'] = df['Quantity'].fillna(1) 

print("\n--- Cleaned Data ---")
print(df)

# 3. DATA VISUALIZATION
plt.figure(figsize=(8, 4))

# Plot: Sales Comparison by Product
sns.barplot(data=df, x='Product', y='Sales', estimator=np.sum, errorbar=None, hue='Product', palette='viridis', legend=False)
plt.title('Total Sales Revenue by Product')
plt.xlabel('Product Item')
plt.ylabel('Total Sales ($)')

plt.tight_layout()
plt.savefig('visual_report.png') # Saves your dashboard output automatically
plt.show()
```
